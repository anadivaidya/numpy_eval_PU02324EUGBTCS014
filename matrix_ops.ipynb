```python
import numpy as np
np.random.seed(42)
A = np.random.randint(1, 10, size=(4, 3))
B = np.random.randint(1, 10, size=(3, 5))

print("Matrix A:\n", A, "\nShape: ", A.shape)
print("\nMatrix B:\n", B, "\nShape: ", B.shape)
```

```python
assert A.shape[1] == B.shape[0], "Inner Dimensions do not match!"
C = A@B
print("Matrix C:\n", C)
print("\nShape of C:", C.shape)
assert C.shape == (4, 5)
```

```python
total_sum = np.sum(C)
row_sums = np.sum(C, axis=1)
col_sums = np.sum(C, axis=0)

print("Total sum of C:", total_sum)
print("Row-wise sums:", row_sums)
print("Column-wise sums:", col_sums)
```
